```python
from dataset_loader import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer
```

```python
loader = HFDatasetLoader(dataset_id="nvidia/Nemotron-Personas-UK", split="train")
df = loader.get_dataframe()
df.head()
```

```python
# Transformación 1: seleccionar columnas clave
columnas_clave = ['age', 'occupation', 'industry', 'country']
df_transformado = df[columnas_clave].copy()

# Transformación 2: filtrar filas por condición
df_transformado = df_transformado[df_transformado['age'] > 30].copy()

# Transformación 3: renombrar columna
df_transformado = df_transformado.rename(columns={'age': 'edad'})

# Transformación 4: crear nueva columna derivada
df_transformado['grupo_edad'] = df_transformado['edad'].apply(
    lambda edad: 'Adulto mayor' if edad >= 60 else 'Adulto'
)

print('Transformaciones aplicadas: selección, filtro, renombre y nueva columna.')
df_transformado.head()
```

```python
summarizer = DatasetSummarizer(df_transformado)
summarizer.generar_resumen(output_file='resumen_personas_uk.txt')
print("Archivo .txt generado exitosamente.")
```
